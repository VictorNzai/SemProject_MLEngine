### Target Variable: Battery Failure Analysis

We've identified `battery_failure` as a key target variable for classification. Let's examine its distribution within the dataset to understand the balance between healthy and failed batteries.

````python
fail_counts = df["battery_failure"].value_counts()
fail_pct = df["battery_failure"].value_counts(normalize=True) * 100

print("Battery Failure Counts:")
display(fail_counts)
print("\nBattery Failure Percentage:")
display(fail_pct.round(2))
````

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 5))
df["battery_failure"].value_counts().plot(kind="bar", color=["steelblue", "indianred"])
plt.xticks([0, 1], ["Healthy (0)", "Failure (1)"], rotation=0)
plt.ylabel("Count")
plt.title("Battery Failure Distribution")
plt.show()
````
