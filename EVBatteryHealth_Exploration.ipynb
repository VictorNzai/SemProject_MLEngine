## Dataset 1: EV Battery Health (20K)

This dataset (`ev battery_failure prediction Dataset(20K).csv`) contains detailed information about electric vehicles and their battery health. It includes vehicle specifications, battery characteristics, usage data, and various sensor readings.

**Key contents include:**
*   **Vehicle Information:** `vehicle_id`, `vehicle_brand`, `vehicle_model`, `vehicle_type`, `manufacturing_year`, `drive_type`.
*   **Battery Specifications:** `battery_manufacturer`, `battery_chemistry`, `battery_capacity_kwh`, `nominal_voltage`, `charge_cycles`.
*   **Usage & Environment:** `odometer_km`, `temperature_c`, `humidity_percent`, `charging_duration_hours`, `driving_conditions`.
*   **Health & Performance Metrics:** `current_v`, `current_a`, `voltage_v`, `power_kw`, `soc_percent`, `soh_percent`, `resistance_ohms`, `cycle_count`, `diagnostic_code`.
*   **Warning & Faults:** `sensor_fault_count`, `BMS_warning_count`, `abnormal_voltage_events`.
*   **Derived Health Scores:** `battery_stress_index`, `aging_score`, `thermal_health_score`, `charging_quality_score`, `driving_stress_score`, `predicted_remaining_life_cycles`.

**Potential Target Variables for ML Application:**
*   **`battery_failure`**: This is a binary variable (0 or 1) indicating whether the battery has failed. It's ideal for a **classification** task, predicting battery failure.
*   **`predicted_remaining_life_cycles`**: This numerical variable predicts the remaining operational life cycles of the battery. It's suitable for a **regression** task, estimating battery lifespan.

## Dataset 2: EV Battery Health (200K) - Detailed Description

This dataset (`ev_battery_failure_dataset.csv`) is a larger collection with 200,000 records detailing electric vehicle battery health, similar in scope to the 20K dataset but with more extensive data. It provides a comprehensive view of various factors influencing battery performance and longevity.

**Key contents include, as per the data dictionary:**
*   **Vehicle Information:** `vehicle_id`, `vehicle_brand`, `vehicle_model`, `vehicle_type`, `manufacturing_year`.
*   **Battery Specifications:** `battery_manufacturer`, `battery_chemistry`, `battery_capacity_kwh`.
*   **Usage & Environment:** `odometer_km`, `temperature_c`, `humidity_percent`, `charging_duration_hours`, `driving_conditions`, `geographic_location`.
*   **Operational Metrics:** `current_v`, `current_a`, `voltage_v`, `power_kw`, `soc_percent`, `soh_percent`, `resistance_ohms`, `cycle_count`, `diagnostic_code`.
*   **Event Counters:** `sensor_fault_count`, `BMS_warning_count`, `abnormal_voltage_events`.
*   **Derived Health Scores:** `battery_stress_index`, `aging_score`, `thermal_health_score`, `charging_quality_score`, `driving_stress_score`.

**Potential Target Variables for ML Application:**
*   **`battery_failure`**: This is explicitly marked as a 'Target' variable in the data dictionary, indicating a binary outcome (1 = battery failure, 0 = healthy). It's suitable for a **classification** model to predict battery failure.
*   **`predicted_remaining_life_cycles`**: Also listed as a 'Target' variable, this estimates the remaining operational life cycles. This is ideal for a **regression** model to predict the remaining lifespan of a battery.

These variables, along with the detailed feature descriptions, provide excellent opportunities for developing predictive models for EV battery health and performance.
